```python
# P4

data1 = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv")
datal
#data1.head()
#data1.describe()
#data1.info()
#data1.tail()
#data1.shape
#data1.columns
#data1.dtypes
```

```python
# Data bukan jan yang >100, bulan feb <300

data1.loc[(data1["jan"] > 100) & (data1["feb"] < 300), ["nama", "jan", "feb"]]
```

```python
#rata rata bulan januari
rata2_jan = data1["jan"].mean()
print("Rata-rata bulan Januari =", rata2_jan)

#variansi bulan januari
variansi_jan = data1["jan"].var()
print("Variansi bulan Januari =", variansi_jan)

#standar deviasi bulan januari
std_jan = data1["jan"].std()
print("Standar deviasi bulan Januari =", std_jan)
```
