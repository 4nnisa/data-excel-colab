```python
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns

bulan =['jan','feb','mar','apr','mei','juni','juli']
penjualan =[150, 200, 100, 250, 125, 299, 300]

plt.figure(figsize=(6,4))

sns.barplot (x=bulan, y=penjualan, palette='muted')

plt.plot(
    range(len(bulan)),
    penjualan,
    color='red',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='garis tren'
)

plt.title('Penjualan per bulan garis tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)
df

df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

df_sorted = df.sort_values('total', ascending=False)

sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')

plt.title('Total Penjualan Per Orang')
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"

df = pd.read_csv(url)

# PENTING: Bersihkan nama kolom dari spasi tersembunyi di awal/akhir
df.columns = df.columns.str.strip()

# 3. Tentukan kolom yang ingin Anda lihat (sesuai permintaan: PAD, Kemiskinan, IPM)
kol_pilihan = ['PAD', 'KEMISKINAN', 'IPM']

# 4. Filter aman: Hanya ambil kolom yang BENAR-BENAR ada di dalam df
# Ini akan mencegah error KeyError jika ada salah ketik
kolom_aman = [kolom for kolom in kol_pilihan if kolom in df.columns]

# 5. Jalankan describe jika kolom ditemukan
if len(kolom_aman) > 0:
    hasil_deskriptif = df[kolom_aman].describe()
    print(hasil_deskriptif)
else:
    print("Tidak ada kolom pilihan yang ditemukan di dataset.")
```
